# Titanic Dataset – Feature Encoding in Python

## Objective
This notebook demonstrates the essential feature encoding techniques used in data preprocessing for machine learning using the Titanic dataset.

## Workflow
- Import the required libraries
- Load and explore the dataset
- Identify categorical features
- Apply Label Encoding
- Apply One-Hot Encoding
- Apply Dummy Encoding
- Apply Ordinal Encoding
- Apply Frequency Encoding
- Apply Manual Mapping

**Dataset:** `Titanic-Dataset.csv`

**Libraries:** Pandas, NumPy, Scikit-learn

```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import (
    LabelEncoder,
    OneHotEncoder,
    OrdinalEncoder
)
```

```python
df = pd.read_csv("Titanic-Dataset.csv")

df.head()
```

```python
df.info()
```

##Label Encoding(Sex)

```python
label_df = df.copy()

le = LabelEncoder()
label_df["Sex"] = le.fit_transform(label_df["Sex"])

label_df[["Sex"]].head()
```

```python
onehot_df = pd.get_dummies(
    df,
    columns=["Embarked"],
    dtype=int
)

onehot_df.head()
```

##One-Hot Encoding(Embarked)

```python
onehot_df = pd.get_dummies(
    df,
    columns=["Embarked"],
    dtype=int
)

onehot_df.head()
```

##Dummy Encoding

```python
dummy_df = pd.get_dummies(
    df,
    columns=["Embarked"],
    drop_first=True,
    dtype=int
)

dummy_df.head()
```

##Ordinal Encoding (Pclass)

```python
ordinal_df = df.copy()

encoder = OrdinalEncoder(categories=[[1, 2, 3]])

ordinal_df["Pclass"] = encoder.fit_transform(
    ordinal_df[["Pclass"]]
)

ordinal_df[["Pclass"]].head()
```

##Frequency Encoding (Embarked)

```python

freq_df = df.copy()

frequency = freq_df["Embarked"].value_counts()

freq_df["Embarked_Freq"] = freq_df["Embarked"].map(frequency)

freq_df[["Embarked", "Embarked_Freq"]].head()
```

##Manual Mapping

```python
map_df = df.copy()

sex_map = {
    "male": 1,
    "female": 0
}

map_df["Sex"] = map_df["Sex"].map(sex_map)

map_df[["Sex"]].head()
```
