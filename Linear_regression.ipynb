# Linear Regression

This project demonstrates Linear Regression using a synthetically generated dataset. It covers data generation, visualization, train-test splitting, model training, prediction, and evaluation using MAE, MSE, RMSE, and R².

```python
from sklearn.datasets import make_regression
import pandas as pd

X, y = make_regression(
    n_samples=100,
    n_features=1,
    noise=10,
    random_state=42
)
```

```python
df = pd.DataFrame({
    "X": X.flatten(),
    "y": y
})

df.head()
```

```python
plt.scatter(df["X"], df["y"], color="pink", alpha=0.7)
plt.xlabel("X")
plt.ylabel("y")
plt.title("X vs y")
plt.show()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df[["X"]],
    df["y"],
    test_size=0.2,
    random_state=42
)
```

```python
df[["X"]]
```

```python
df["y"]
```

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
model.fit(X_train, y_train)
```

```python
print("Slope:", model.coef_)
print("Intercept:", model.intercept_)
```

```python
new_X = [[2]]

prediction = model.predict(new_X)

print(prediction)
```

```python
y_pred = model.predict(X_test)
```

```python
print(y_pred[:5])
```

```python
comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

print(comparison.head(10))
```

```python
y_pred = model.predict(X_test)

comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

print(comparison.head(10))
```

```python
from sklearn.metrics import mean_absolute_error
mae = mean_absolute_error(y_test, y_pred)

print("MAE:", mae)
```

```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_test, y_pred)

print("MSE:", mse)
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np

mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

print("MSE:", mse)
print("RMSE:", rmse)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)

print("R²:", r2)
```

```python
plt.scatter(X_test, y_test, color="pink")
plt.plot(X_test, y_pred, color="purple")
plt.xlabel("X")
plt.ylabel("y")
plt.title("Linear Regression 💗")
plt.show()
```
